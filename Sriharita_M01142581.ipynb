```python
# Create a cleaned copy of the original dataset
df_clean = df.drop_duplicates().copy()

# Compare dataset size before and after removing duplicates
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

# Confirm that no exact duplicate rows remain
print("Remaining duplicate rows:", df_clean.duplicated().sum())

# Check missing billing amounts
print("Missing billing amounts:", df_clean["Billing Amount"].isnull().sum())
```

```python
# Convert admission and discharge dates into datetime format
df_clean["Date of Admission"] = pd.to_datetime(
    df_clean["Date of Admission"],
    format="%d/%m/%Y",
    errors="coerce"
)

df_clean["Discharge Date"] = pd.to_datetime(
    df_clean["Discharge Date"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Check the converted dates and data types
print(df_clean[["Date of Admission", "Discharge Date"]].head())

print("\nDate column data types:")
print(df_clean[["Date of Admission", "Discharge Date"]].dtypes)

# Check whether any dates could not be converted
print("\nMissing admission dates:",
      df_clean["Date of Admission"].isnull().sum())

print("Missing discharge dates:",
      df_clean["Discharge Date"].isnull().sum())
```

```python
# Calculate the number of days each patient stayed
df_clean["Length of Stay"] = (
    df_clean["Discharge Date"] - df_clean["Date of Admission"]
).dt.days

# Display the first 10 records
display(
    df_clean[
        ["Date of Admission", "Discharge Date", "Length of Stay"]
    ].head(10)
)

# Check the summary statistics
print("Length of Stay Summary:")
print(df_clean["Length of Stay"].describe())
```

```python
print("Negative length of stay:",
      (df_clean["Length of Stay"] < 0).sum())

print("Zero-day stays:",
      (df_clean["Length of Stay"] == 0).sum())

print("Missing length of stay:",
      df_clean["Length of Stay"].isnull().sum())
```

```python
# Check the data types of all columns
print("DATA TYPES")
print(df_clean.dtypes)

# Check summary statistics for numerical columns
print("\nNUMERICAL SUMMARY")
display(df_clean.describe())
```

```python
# Count negative billing amounts
negative_bills = df_clean[df_clean["Billing Amount"] < 0]

print("Negative billing amounts:", len(negative_bills))

# Display the records with negative bills
display(
    negative_bills[
        ["Age", "Medical Condition", "Hospital",
         "Billing Amount", "Admission Type", "Length of Stay"]
    ].head(10)
)

# Count zero billing amounts
print("Zero billing amounts:",
      (df_clean["Billing Amount"] == 0).sum())

# Check missing billing records
display(
    df_clean[df_clean["Billing Amount"].isnull()]
)
```

```python
# Summary of negative billing amounts
print("Negative billing summary:")
print(df_clean.loc[
    df_clean["Billing Amount"] < 0, "Billing Amount"
].describe())

# Overall billing summary
print("\nOverall billing summary:")
print(df_clean["Billing Amount"].describe())

# Percentage of records with negative bills
negative_percentage = (
    (df_clean["Billing Amount"] < 0).sum()
    / df_clean["Billing Amount"].notna().sum()
) * 100

print(f"\nPercentage of negative bills: {negative_percentage:.2f}%")
```

```python
# Audit the cleaned dataset for invalid values and formatting issues

print("Dataset shape:", df_clean.shape)

# 1. Check duplicate records
print("\nDuplicate rows:", df_clean.duplicated().sum())

# 2. Check missing values
print("\nMissing values by column:")
print(df_clean.isnull().sum())

# 3. Check potentially implausible ages
print("\nInvalid ages (below 0 or above 120):")
print(((df_clean["Age"] < 0) | (df_clean["Age"] > 120)).sum())

# 4. Check invalid length of stay
print("\nNegative stays:", (df_clean["Length of Stay"] < 0).sum())
print("Zero-day stays:", (df_clean["Length of Stay"] == 0).sum())

# 5. Check discharge dates before admission dates
print(
    "\nDischarge before admission:",
    (df_clean["Discharge Date"] < df_clean["Date of Admission"]).sum()
)

# 6. Check negative billing values
print("\nNegative billing amounts:")
print((df_clean["Billing Amount"] < 0).sum())

# 7. Check inconsistent whitespace or casing in text columns
print("\nText formatting audit:")

for col in df_clean.select_dtypes(include="object").columns:
    values = df_clean[col].dropna().astype(str)

    whitespace_count = (
        values.str.strip().ne(values).sum()
    )

    casing_variants = (
        values.str.strip().str.casefold().nunique()
        < values.nunique()
    )

    if whitespace_count > 0 or casing_variants:
        print(
            f"{col}: leading/trailing spaces = {whitespace_count}, "
            f"possible casing variants = {casing_variants}"
        )
```

```python
# Find names that differ only by capitalisation
name_variants = df_clean.copy()

name_variants["Name_Normalised"] = (
    name_variants["Name"].str.strip().str.casefold()
)

name_groups = (
    name_variants.groupby("Name_Normalised")["Name"]
    .agg(lambda x: sorted(x.unique()))
)

name_groups = name_groups[
    name_groups.apply(lambda x: len(x) > 1)
]

print("Number of names with possible casing variants:", len(name_groups))
print(name_groups.head(20))
```

```python
# Count distinct original names and normalised names
original_unique_names = df_clean["Name"].nunique()

normalised_unique_names = (
    df_clean["Name"].str.strip().str.casefold().nunique()
)

print("Distinct original names:", original_unique_names)
print("Distinct normalised names:", normalised_unique_names)
print(
    "Difference after normalisation:",
    original_unique_names - normalised_unique_names
)
```
